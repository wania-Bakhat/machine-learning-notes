```python
from sklearn.datasets import make_blobs
#it creates bolbs of dataset mean creating blobs/clusters of datapoints
from sklearn.datasets import make_moons #from this you can make certain shapes of scatterplot
```

```python
X,y=make_moons(noise=0.1,random_state=42)
#this randomstate=42 ensures we always get the same dataset always its like a seed
```

```python
X,y=make_blobs(n_samples=5000,centers=5)
#5000 instances and 5 clusters/blobs we can pass a parameter i.e feature you can do feature=5 for making data 5 dimensional
X
y
```
