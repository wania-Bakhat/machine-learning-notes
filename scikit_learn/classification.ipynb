The goal is:
Can a machine predict whether a tumor is malignant (cancerous) or benign (non-cancerous)?
This is called a classification problem because we're predicting a category (class), not a number.
no matter what classifier/algorithm you use the procedure or steps are always the same:
1.you import a classifier and create its instance
2.fit the model 
3.evaluate the model (score it)
4.prediction

Every ML project usually follows these steps:

Collect data
      ↓
Split data
      ↓
Scale data (sometimes)
      ↓
Create model
      ↓
Train model
      ↓
Evaluate model
      ↓
Predict new data

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X,y=load_breast_cancer(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier #This imports the K-Nearest Neighbors (KNN) algorithm
''' 
The idea is simple:
Imagine a new patient.The algorithm looks at the K nearest patients in the training data.
Example
New Patient
Nearby patients:
Cancer
Cancer
Cancer
Benign
Cancer
Most neighbors have cancer.
Prediction:
Cancer
It predicts based on the majority vote of nearby examples.
'''
clf=KNeighborsClassifier()
clf.fit(X_train_scaled,y_train)
```

**fit() = Learn from the data.**
For a machine learning model:
clf.fit(X_train, y_train)
means:
“Hey model, look at these examples (X_train) and their correct answers (y_train), and learn the pattern.”
**What does transform() do?**
transform() = Apply a transformation using what was already learned.
Example with StandardScaler:
scaler.fit(X_train)
The scaler learns:
Mean (average) of each feature
Standard deviation of each feature
scaler.transform(X_test)
uses those learned values to scale new data.
It does not learn anything new.
**3. What does fit_transform() do?**
fit_transform() = fit() + transform() together.
scaler.fit_transform(X_train)
This is equivalent to:
So it:
Learns from X_train
Immediately transforms X_train

```python
clf.score(X_test_scaled,y_test) #This evaluates the model.
```

```python
single_instance=X_test_scaled[0] # always use scaled data or else prediction would be wrong
''' 
This selects just one patient from the testing data. Instead of predicting everyone,
we predict one patient.
'''
single_instance
```

```python
clf.predict([single_instance]) #Predict the class of this one patient.
```
