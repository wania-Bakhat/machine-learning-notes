THIS NOTEBOOK DISCUSSES PREPROCESSING 

The idea of preprocessing is that you have the data that is not perfect to use rightaway with your model.
-> Some of the features might not be numerical.
-> some of the features might not be scaled yet.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler #it is helpful for scaling data
```

```python
X,y=load_iris(return_X_y=True)
X #we see the data is not scaled not normalized just the raw centimeter values
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
#now when we use it on testing data we dont fit it we just transform
X_test_scaled=scaler.transform(X_test)
''' 
why we do fit transform?
because the scaler has to learn certain values. what it does is it takes a sample and mean from the sample and the standard deviation or varience from sample
then it perform some calculations based on the values it learned from training data and the reason we use training data is because we are learning these values 
and doing the calculation
'''
```

STANDARD SCALER:
a standard scaler basically subtracts the mean from data and divides by the standard deviation. By doing this we zero center teh Data and and bring it to unit variance.

```python
#doing same thing standard scaler does by numpy to prove:
import numpy as np
(X_train-np.mean(X_train,axis=0))/np.std(X_train,axis=0)
```

MIN-MAX SCALING: you press all values to be between 0-1


```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
#doing same thing min-max scaler does by numpy to prove:
import numpy as np
X_min=np.min(X_train,axis=0)
X_max=np.max(X_train,axis=0)
(X_train - X_min) / (X_max - X_min)
```
