```python
from sklearn.datasets import fetch_openml #this allow us to load datasetd from openML 
from sklearn.preprocessing import OrdinalEncoder 
```

```python
data=fetch_openml('car',as_frame=True).frame
```

all ML model works on numbers so we have to convert all the non numerical columns into numerical columns

**METHOD:01 ordinal encoder:**
what ordinal encoder does is according to the index of list elements in categories like [small,medium,big] small->0, medium->1 , big->2 instead of the values like small,medium big (which are non numeric) in 'lug_boot' column now we will have 0 ,1 ,2 0 for small 1 for medium and 2 for big

```python
columns_to_encode=['lug_boot','safetly']
encoder=OrdinalEncoder(
    categories=[
        ['small','med','big'],
        ['low','med','high']
    ]
)
data[columns_to_encode]=encoder.fit_transform(data[columns_to_encode])
```

```python
#inverse transform turns the columns back to what they were initially
encoder.inverse_transform(data[columns_to_encode])
```

**METHOD:02 one hot encoder:**
it turns multiple values for a column into multiple columns that can be 0 or 1 (like a boolean value)

```python
data=fetch_openml('adult',as_frame=True).frame
```

```python
from sklearn.preprocessing import OneHotEncoder
''' 
we have a column occupation which has different values like
 adm-clerical exec-managerial etc we will make cindividual columns for each such value )
'''
data.occupation.value_counts()
```

```python
import pandas as pd
encoder=OneHotEncoder(handle_unknown='ignore',sparse=False)
encoded_values=encoder.fit_transform(data['occupation','race'])
new_cols=encoder.get_feature_names_out(data['occupation','race'])
```

```python
df_encoded=pd.DataFrame(encoded_values,columns=new_cols,index=data.index)
#merging:
data_final=pd.concat(
    #drop the 2 columns and merge with encoded data frame on axis 1 (means column)
    [data.drop(columns=['occupation','race']),df_encoded],
    axis=1
)
```
