```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
data=load_iris()
X,y=data.data, data.target
```

```python
X,y=load_iris(return_X_y=True)
```

**PROBLEM OF OVERFITTING DATA:**
we train and evaluate a model on different data in machine learning. suppose we are learning about cats and dogs and i show you the pictures of cats and dogs and you just learn to distinguish between them by looking at the pictures not by learning about their different features you see thats the problem because maybe you don't generalize well and you just got the training phase well by memorizing pictures but if i show you pictures you have never seen before maybe you fail completely because you never learned to generalize so what we do in machine learning is train on 1 portion of data and we evalute on unseen portion of data.

```python
#splitting data in training and testing set:
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
#test size 0.2 means 20% data us used for testing and remaining 80% is used for training
```

we will now visualize distribution of classification of labels is it equal or very different because the problem is if i have a training set where 1 or 2 classes are present and 3rd one is completely absent or present in small amount while in testing set its completely different in this case you can't say your evaluation is completely meaningfull because you are working with different datasets 

```python
import numpy as np
import matplotlib.pyplot as plt
counts=np.bincount(y_train)
positions=np.arange(3)
plt.bar(positions,counts)
plt.xticks(positions,data.target_names)
```

```python
import numpy as np
import matplotlib.pyplot as plt
counts=np.bincount(y_test)
positions=np.arange(3)
plt.bar(positions,counts)
plt.xticks(positions,data.target_names)
''' 
if the training data and testing data are completely opposite (which doesnot happen everytime necessarily but it can happen sometimes )
then the graphs of y_train and y_test would be completely opposite
'''
```

**stratified shuffle split**
to eliminate this probelm we will use another model from sklearn that is stratified shuffle split instead of train and test split 
what this does is it makes sure to give us similar distribution or same ratio of different classes in training and testing data

```python
from sklearn.model_selection import StratifiedShuffleSplit
split=StratifiedShuffleSplit(n_splits=1,test_size=0.2)
#split=1 means we only need a single split
for train_idx, test_idx in split.split(X,y):
    X_train,X_test=X[train_idx],X[test_idx]
    y_train,y_test=y[train_idx],y[test_idx]

```

now you can see by running the below code portions that all classes have equal distribution or ratio in both graphs of training and testing data

```python
import numpy as np
import matplotlib.pyplot as plt
counts=np.bincount(y_train)
positions=np.arange(3)
plt.bar(positions,counts)
plt.xticks(positions,data.target_names)
```

```python
import numpy as np
import matplotlib.pyplot as plt
counts=np.bincount(y_test)
positions=np.arange(3)
plt.bar(positions,counts)
plt.xticks(positions,data.target_names)
```
